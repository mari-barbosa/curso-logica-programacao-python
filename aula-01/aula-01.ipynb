## A função print( )

Print é a função em Python que usamos para exibir algo na tela.

Sua estrutura básica é: print(O que queremos exibir).

```python
print (Olá Mundo!)
```

```python
print ("Olá Mundo!")
```

Nós também podemos usar vários prints em sequência:

```python
print("Iniciando o curso de Python...")
print("Esta é a minha primeira instrução.")
print("Programar é mais fácil do que parece!")
```
