## Por que Python?

- Python é uma das melhores linguagens de programação para iniciantes.

- Ela foi criada em 1989 e o seu nome é uma homenagem à série *Monty Python.*

- Se parece muito com o inglês escrito.

- É a base das inteligências artificiais mais comuns (GPTs, por exemplo).

- É usada por muitos profissionais onde a atividade profissional principal não é a programação

## Jupyter Notebook e o Google Colab

O Jupyter Notebook é uma aplicação web de código aberto que permite criar e compartilhar documentos interativos combinando código executável, textos explicativos, fórmulas e gráficos em um só lugar.

O nome “Jupyter” deriva das principais linguagens de programação suportadas: Julia, Python e R.

O Google Colaboratory, ou Google Colab, é uma plataforma gratuita que fornece um ambiente de Jupyter Notebook hospedado na nuvem que permite escrever e executar código Python.

**O único requisito é ter uma conta Google**
