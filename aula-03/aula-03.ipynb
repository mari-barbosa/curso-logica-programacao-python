## Recebendo Dados com input( )

Se o print( ) é a "impressora" do programa, o input( ) é o "microfone".

Ele pausa a execução do código e espera o usuário digitar algo e apertar Enter.

```python
nome = input("Qual é o seu nome? ")
print(f"Prazer em te conhecer, {nome}!")
```

## A Armadilha do input( ) e Conversão de Tipos

Tudo o que o input( ) lê entra no programa como Texto (string), mesmo que o usuário digite apenas números.

```python
idade = input("Digite sua idade: ")
preco = input("Digite o preço do item: R$ ")
```

```python
print(type(idade))
print(type(preco))
```

**A Solução**: antes do input, precisamos digitar qual tipo de dado queremos que aquela variável guarde.

```python
idade = int(input("Digite sua idade: "))
preco = float(input("Digite o preço do item: R$ "))
```

```python
print(type(idade))
print(type(preco))
```
