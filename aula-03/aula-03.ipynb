## Recebendo Dados com input()

Se o print() é a "boca" do programa, o input() são os "ouvidos". Ele pausa a execução do código e espera o usuário digitar algo e apertar Enter.

```python
nome = input("Qual é o seu nome? ")
print(f"Prazer em te conhecer, {nome}!")
```

## A Armadilha do input() e Conversão de Tipos

tudo o que o input() lê entra no programa como Texto (string), mesmo que o usuário digite apenas números.

```python
num1 = input("Digite um número: ")  # Usuário digita 5
num2 = input("Digite outro: ")     # Usuário digita 3
print(num1 + num2)                 # Resultado: "53" (concatenação, não soma!)
```

A Solução (Casting): Mostrar como envelopar o input() com int() ou float() para converter o dado recebido.

```python
idade = int(input("Digite sua idade: "))
preco = float(input("Digite o preço do item: R$ "))
```

## Exercícios Práticos

### Exercício 1: Calculadora de Média (Fixação de Casting e Operadores)

Crie um programa que peça ao usuário duas notas de um aluno, calcule a média aritmética entre elas e exiba o resultado final formatado.

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print(f"A média final do aluno é: {media:.1f}")
```

Exercício 2: Conversor de Moedas (Aplicações Práticas)Enunciado: Escreva um programa que peça quanto dinheiro em reais (R$) o usuário tem na carteira e calcule quantos dólares ($) ele pode comprar. *(Considere a cotação fixa de US$ 1.00 = R$ 5.00 para fins didáticos)*.

```python
reais = float(input("Quanto dinheiro você tem na carteira? R$ "))
cotacao = 5.00

dolares = reais / cotacao

print(f"Com R$ {reais:.2f}, você pode comprar US$ {dolares:.2f}")
```

Exercício 3: O Desafio da Parcela (Raciocínio Lógico e Resto)

    Enunciado: Uma loja vende produtos em 3 vezes sem juros. Faça um programa que receba o valor total de uma compra, calcule o valor de cada parcela inteira e mostre quanto "sobra" de troco/centavos no resto da divisão se a conta não for exata.

```python
valor_compra = float(input("Digite o valor total da compra: R$ "))

parcela = valor_compra // 3
resto = valor_compra % 3

print(f"Sua compra será parcelada em 3x de R$ {parcela:.2f}")
print(f"Sobram R$ {resto:.2f} de ajuste de centavos no total.")
```
