## Manipulando Listas

### Função len( )

Retorna o tamanho total da lista (quantidade de itens)

```python
len(frutas)
```

### Método .append( )

Adiciona um novo elemento ao final da lista

```python
frutas.append("jaca")
```

```python
print(frutas)
```

### Método .remove( )

Remove um elemento da lista

```python
frutas.remove("jaca")
```

```python
print(frutas)
```

### Percorrendo a lista com for( )

```python
compras = ["Pão", "Leite", "Café"]

for item in compras:
    print(f"- {item}")
```

### Exercício 2: Cadastro de Alunos Dinâmico

Crie um programa que permita ao usuário cadastrar nomes de alunos em uma lista.

O programa deve continuar pedindo nomes até que o usuário digite "fim".

No final, exiba a quantidade de alunos cadastrados e a lista completa em ordem de cadastro.

```python
alunos = []
nome = ""

print("=== CADASTRO DE ALUNOS ===")
while nome.lower() != "fim":
    nome = input("Digite o nome do aluno (ou 'fim' para encerrar): ")
    if nome.lower() != "fim":
        alunos.append(nome)

print(f"\nTotal de alunos cadastrados: {len(alunos)}")
print("Lista de Alunos:")
for aluno in alunos:
    print(f"• {aluno}")
```

### Exercício 3: Analisador de Números

Escreva um programa que leia 5 números inteiros do teclado, guarde-os em uma lista e depois exiba:

- Todos os números digitados.

- O maior número da lista (Dica: use a função max(lista) ou crie uma lógica com for).

- A soma de todos os valores da lista (Dica: use a função sum(lista) para essa linha de código).

```python
numeros = []

for i in range(1, 6):
    num = int(input(f"Digite o {i}º número inteiro: "))
    numeros.append(num)

print(f"\nNúmeros digitados: {numeros}")
print(f"Maior número: {max(numeros)}")
print(f"Soma de todos os números: {sum(numeros)}")
```
