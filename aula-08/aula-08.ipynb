# Aula 08: Guardando Listas de Coisas

O Conceito de Lista

Analogia do Armário de Gavetas / Lista de Compras: Em vez de ter variáveis separadas item1, item2, item3, criamos uma única "pasta" que guarda todos os itens em ordem.

Sintaxe Básica: As listas em Python usam colchetes [] e separam os elementos por vírgula.

```python
frutas = ["Maçã", "Banana", "Laranja", "Uva"]
```

A Regra do Índice Zero

A Posição dos Elementos: O Python começa a contar as posições (índices) do número 0.

    frutas[0] -> "Maçã"

    frutas[1] -> "Banana"

    frutas[-1] -> "Uva" (índices negativos acessam de trás para frente!)

Manipulando Listas

len(lista): Retorna o tamanho total (quantidade de itens).

lista.append(item): Adiciona um novo elemento ao final da lista.

lista.remove(item) ou del lista[indice]: Remove um elemento.

Percorrendo a lista com for:

```python
compras = ["Pão", "Leite", "Café"]

# Forma direta (item por item)
for item in compras:
    print(f"- {item}")
```

## Exercícios Práticos

Exercício 1: Acessando e Alterando Elementos (Fixação de Índices)

    Enunciado: Crie uma lista com 5 nomes de cidades. Em seguida:

        Imprima a primeira e a última cidade da lista.

        Substitua a terceira cidade por uma nova cidade digitada pelo usuário.

        Exiba a lista atualizada.

```python
cidades = ["São Paulo", "Rio de Janeiro", "Curitiba", "Salvador", "Manaus"]

print(f"Primeira cidade: {cidades[0]}")
print(f"Última cidade: {cidades[-1]}")

cidades[2] = input("Digite o nome de uma nova cidade para substituir Curitiba: ")

print("\nLista de cidades atualizada:")
print(cidades)
```

Exercício 2: Cadastro de Alunos Dinâmico (Uso do append e while)

    Enunciado: Crie um programa que permita ao usuário cadastrar nomes de alunos em uma lista. O programa deve continuar pedindo nomes até que o usuário digite "fim". No final, exiba a quantidade de alunos cadastrados e a lista completa em ordem de cadastro.

```python
alunos = []
nome = ""

print("=== CADASTRO DE ALUNOS ===")
while nome.lower() != "fim":
    nome = input("Digite o nome do aluno (ou 'fim' para encerrar): ")
    if nome.lower() != "fim":
        alunos.append(nome)

print(f"\nTotal de alunos cadastrados: {len(alunos)}")
print("Lista de Alunos:")
for aluno in alunos:
    print(f"• {aluno}")
```

Exercício 3: Analisador de Números (Lógica com Listas e for)

    Enunciado: Escreva um programa que leia 5 números inteiros do teclado, guarde-os em uma lista e depois exiba:

        Todos os números digitados.

        O maior número da lista (Dica: use a função max(lista) ou crie uma lógica com for).

        A soma de todos os valores da lista.

```python
numeros = []

for i in range(1, 6):
    num = int(input(f"Digite o {i}º número inteiro: "))
    numeros.append(num)

print(f"\nNúmeros digitados: {numeros}")
print(f"Maior número: {max(numeros)}")
print(f"Soma de todos os números: {sum(numeros)}")
```
