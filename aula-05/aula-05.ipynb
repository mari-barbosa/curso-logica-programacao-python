## Testando Múltiplas Condições com elif

### A limitação do if/else simples

Se precisamos verificar 3 ou mais cenários (ex: Menor de idade, Adulto, Idoso), aninhar vários if/else fica confuso.

```python
idade = 72

if idade < 18:
    if idade >= 0:
        if idade < 13:
            categoria = "Criança"
        else:
            categoria = "Adolescente"
    else:
        categoria = "Idade inválida"
else:
    if idade < 65:
        if idade < 30:
            categoria = "Jovem"
        else:
            if idade < 50:
                categoria = "Jovem Adulto"
            else:
                categoria = "Adulto"
    else:
        if idade < 80:
            categoria = "Idoso"
        else:
            if idade < 100:
                categoria = "Ancião"
            else:
                categoria = "Centenário"
print(categoria)
```

## Cláusula elif

- elif é a abreviação de "else if" (senão se).

- Ele só é testado se o if anterior (ou o elif acima) for falso.

```python
idade = 72

if idade < 12:
    print("Criança")
elif idade < 18:
    print("Adolescente")
elif idade < 60:
    print("Adulto")
else:
    print("Idoso")
```

### Exercício 1: Classificador de Notas Escolares

Escreva um programa que leia a nota final de um aluno (de 0 a 10) e exiba a situação correspondente:

- Nota de 7.0 a 10.0: Aprovado

- Nota de 5.0 a 6.9: Em Recuperação

- Nota abaixo de 5.0: Reprovado

```python
nota = float(input("Digite a nota do aluno (0 a 10): "))

if nota >= 7.0:
    print("Situação: APROVADO! Parabéns.")
elif nota >= 5.0:
    print("Situação: EM RECUPERAÇÃO. Estude mais um pouco.")
else:
    print("Situação: REPROVADO. Tente novamente no próximo período.")
```

### Exercício 2: Validador de Empréstimo Bancário

Um banco aprova empréstimos apenas se a pessoa tiver idade mínima de 21 anos E uma renda mensal de pelo menos R$ 2.500,00. 

Crie um programa que receba esses dois dados do usuário e informe se o empréstimo foi "Aprovado" ou "Negado".

```python
idade = int(input("Digite a sua idade: "))
renda = float(input("Digite a sua renda mensal: R$ "))

if idade >= 21 and renda >= 2500.0:
    print("Empréstimo APROVADO com sucesso!")
else:
    print("Empréstimo NEGADO. Requisitos não atingidos.")
```

### Exercício 3: Sistema de Desconto no Cinema

No cinema local, o ingresso custa R$ 30,00. Têm direito a 50% de desconto pessoas que forem estudantes OU pessoas com 60 anos ou mais.

Faça um programa que leia a idade e pergunte se o usuário é estudante (com "S" ou "N") e exiba o preço final a pagar.


```python
idade = int(input("Digite sua idade: "))
resposta_estudante = input("Você é estudante? (S/N): ").strip().upper()

estudante = (resposta_estudante == "S")

if idade >= 60 or estudante:
    print("Você tem direito à meia-entrada! Valor: R$ 15,00")
else:
    print("Valor do ingresso inteiro: R$ 30,00")
```
