## Testando Múltiplas Condições com elif

A limitação do if/else simples: Se precisamos verificar 3 ou mais cenários (ex: Menor de idade, Adulto, Idoso), aninhar vários if/else fica confuso.

Sintaxe do elif: Abreviação de "else if" (senão se). Ele só é testado se o if anterior (ou o elif acima) for falso.

```python
idade = int(input("Digite sua idade: "))

if idade < 12:
    print("Criança")
elif idade < 18:
    print("Adolescente")
elif idade < 60:
    print("Adulto")
else:
    print("Idoso")
```

## Combinando Condições com Operadores Lógicos

and (E): Exige que todas as condições sejam verdadeiras.

    Exemplo da vida real: Para entrar na festa, precisa ter ingresso E ter mais de 18 anos.

or (OU): Exige que pelo menos uma das condições seja verdadeira.

    Exemplo da vida real: Tem direito a meia-entrada quem é estudante OU quem é idoso.

not (NÃO): Inverte o valor lógico (o que é True vira False e vice-versa).

```python
# Exemplo combinando operadores
tem_ingresso = True
idade = 20

if tem_ingresso and idade >= 18:
    print("Entrada liberada!")
else:
    print("Entrada proibida.")
```

## Exercícios Práticos

Exercício 1: Classificador de Notas Escolares (Uso do elif)

    Enunciado: Escreva um programa que leia a nota final de um aluno (de 0 a 10) e exiba a situação correspondente:

        Nota de 7.0 a 10.0: Aprovado

        Nota de 5.0 a 6.9: Em Recuperação

        Nota abaixo de 5.0: Reprovado

```python
nota = float(input("Digite a nota do aluno (0 a 10): "))

if nota >= 7.0:
    print("Situação: APROVADO! Parabéns.")
elif nota >= 5.0:
    print("Situação: EM RECUPERAÇÃO. Estude mais um pouco.")
else:
    print("Situação: REPROVADO. Tente novamente no próximo período.")
```

Exercício 2: Validador de Empréstimo Bancário (Uso do and)

    Enunciado: Um banco aprova empréstimos apenas se a pessoa tiver idade mínima de 21 anos E uma renda mensal de pelo menos R$ 2.500,00. Crie um programa que receba esses dois dados do usuário e informe se o empréstimo foi "Aprovado" ou "Negado".

```python
idade = int(input("Digite a sua idade: "))
renda = float(input("Digite a sua renda mensal: R$ "))

if idade >= 21 and renda >= 2500.0:
    print("Empréstimo APROVADO com sucesso!")
else:
    print("Empréstimo NEGADO. Requisitos não atingidos.")
```

Exercício 3: Sistema de Desconto no Cinema (Uso do or)

    Enunciado: No cinema local, o ingresso custa R$ 30,00. Têm direito a 50% de desconto pessoas que forem estudantes OU pessoas com 60 anos ou mais. Faça um programa que leia a idade e pergunte se o usuário é estudante (com "S" ou "N") e exiba o preço final a pagar.

```python
idade = int(input("Digite sua idade: "))
resposta_estudante = input("Você é estudante? (S/N): ").strip().upper()

eh_estudante = (resposta_estudante == "S")

if idade >= 60 or eh_estudante:
    print("Você tem direito à meia-entrada! Valor: R$ 15,00")
else:
    print("Valor do ingresso inteiro: R$ 30,00")
```
