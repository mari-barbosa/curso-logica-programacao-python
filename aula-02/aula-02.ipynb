### Regras de Ouro para Nomes de Variáveis (em Python)

- Não pode começar com números (1nome ❌ -> nome1 ✅).

- Não pode ter espaços (nome aluno ❌ -> nome_aluno ✅ - estilo snake_case).

- Python diferencia maiúsculas de minúsculas (Numero é diferente de numero).

- Escolha nomes explicativos (idade é melhor do que x).

- Não pode ser uma "palavra reservada" (palavras especiais utilizadas pelo Python para outras funcionalidades).
<br>

## Números Inteiros (int)

Números sem casas decimais.

````python
10
````

## Números Decimais (float)

Em programação, chamamos os números que possuem casas decimais de "números de ponto flutuante".

Em Python, não seu usa " , " para escrever um número decimal e sim " . "

````python
14.7
````

````python
14,7
````

## Verdadeiro/Falso (bool): Valores lógicos: True ou False (sempre com a primeira letra maiúscula).

Lógica Booleana -> Um valor só pode ser verdadeiro (True) ou falso (False). É utilizado para testes lógicos.

````python
7 == 5
````

````python
8 == 8
````
