## O sinal de " = " em programação

O sinal de igual (=) em programação não significa "igualdade matemática", mas sim atribuição ("guarde o que está à direita dentro da caixa da esquerda")

### Os 4 Tipos Básicos de Dados

Texto (string): Sempre entre aspas. Ex: "Python", "123" (se está entre aspas, é texto!).

Número Inteiro (int): Números sem vírgula/ponto. Ex: 10, -5, 2026.

Número Decimal (float): Números com ponto decimal (padrão americano). Ex: 1.75, 49.90.

Verdadeiro/Falso (bool): Valores lógicos: True ou False (sempre com a primeira letra maiúscula).

```python
nome = "Mari"       # String
idade = 22          # Int
altura = 1.70       # Float
estudante = True    # Bool
```

## Regras de Ouro para Nomes de Variáveis

Não pode começar com números (1nome ❌ -> nome1 ✅).

Não pode ter espaços (nome aluno ❌ -> nome_aluno ✅ - estilo snake_case).

Python diferencia maiúsculas de minúsculas (Nome é diferente de nome).

Escolha nomes explicativos (idade é melhor do que x).

## Exibindo Variáveis no print() com f-strings

```python
nome = "Mari"
idade = 22
print(f"Olá, meu nome é {nome} e tenho {idade} anos.")
```
