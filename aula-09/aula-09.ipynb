## Criando a Primeira Função com def

A palavra reservada **def** (de define), indica que estamos iniciando a criação de uma função.

Logo depois, declaramos o nome da função (lembre-se das regras para nomes de variáveis, é bem parecido).

Depois digitamos os parênteses ( ) e os dois pontos :

```python
def dar_boas_vindas():
    print("**************************")
    print("  BEM-VINDO AO SISTEMA!   ")
    print("**************************")

dar_boas_vindas()
```

### Passando Informações com Parâmetros

```python
def saudar_usuario(nome):
    print(f"Olá, {nome}! Seja muito bem-vindo.")

saudar_usuario("Mari")
saudar_usuario("Carlos")
```

```python
def saudar_usuario(nome):
    print(f"Olá, {nome}! Seja muito bem-vindo.")

saudar_usuario(input("Digite seu nome: "))
```

### A Importância do return

Funções podem "retornar" uma resposta (valor de retorno).

Para "capturaremos" essa resposta utilizamos a instrução **return**

```python
def somar(a, b):
    resultado = a + b

total = somar(10, 15)
```

```python
total
```

```python
def somar(a, b):
    resultado = a + b
    return resultado

# Guardando o retorno da função em uma variável
total = somar(10, 15)
```

```python
total
```

## Exercícios Práticos

### Exercício 1: Validador de Maioridade

Crie uma função chamada eh_maior_de_idade(idade) que receba uma idade como parâmetro e retorne True se a idade for maior ou igual a 18, ou False caso contrário. No programa principal, peça a idade do usuário e use a função para exibir uma mensagem correspondente.

```python
def eh_maior_de_idade(idade):
    if idade >= 18:
        return True
    else:
        return False

# Programa Principal
idade_usuario = int(input("Digite sua idade: "))

if eh_maior_de_idade(idade_usuario):
    print("Acesso liberado: Você é maior de idade.")
else:
    print("Acesso negado: Você é menor de idade.")
```

### Exercício 2: Calculadora de Área de Retângulo

Escreva uma função chamada calcular_area(largura, comprimento) que receba as dimensões de um terreno retangular e retorne a sua área total em metros quadrados.

```python
def calcular_area(largura, comprimento):
    area = largura * comprimento
    return area

# Programa Principal
print("=== CÁLCULO DE TERRENO ===")
larg = float(input("Digite a largura do terreno (m): "))
comp = float(input("Digite o comprimento do terreno (m): "))

area_terreno = calcular_area(larg, comp)
print(f"A área total do terreno é de {area_terreno:.2f} m².")
```

### Exercício 3: Módulos da Calculadora

Crie quatro funções: somar(a, b), subtrair(a, b), multiplicar(a, b) e dividir(a, b). A função de divisão deve verificar se o divisor é zero antes de calcular para evitar erros. Em seguida, peça dois números ao usuário e execute as quatro operações usando suas funções.

```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return "Erro: Divisão por zero não permitida!"
    return a / b

# Programa Principal
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))

print("\n=== RESULTADOS ===")
print(f"Soma: {somar(n1, n2)}")
print(f"Subtração: {subtrair(n1, n2)}")
print(f"Multiplicação: {multiplicar(n1, n2)}")
print(f"Divisão: {dividir(n1, n2)}")
```
