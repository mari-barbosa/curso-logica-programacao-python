# Aula 09: Reutilizando Código com Funções

## O Conceito de Função

Analogia do Eletrodoméstico / Liquidificador: Uma função é como um liquidificador. Você passa os ingredientes (parâmetros de entrada), ele executa um processo interno e entrega uma vitamina pronta (retorno/resultado).

Por que usar funções? Evita duplicar código (princípio DRY: Don't Repeat Yourself), facilita a correção de bugs e deixa o programa principal mais limpo.

## Criando a Primeira Função com def

Explicar a palavra reservada def (de define), o nome da função, os parênteses () e os dois pontos :.

```python
def dar_boas_vindas():
    print("**************************")
    print("  BEM-VINDO AO SISTEMA!   ")
    print("**************************")

# Chamando a função para executar
dar_boas_vindas()
```

Passando Informações com Parâmetros

```python
def saudar_usuario(nome):
    print(f"Olá, {nome}! Seja muito bem-vindo.")

saudar_usuario("Mari")
saudar_usuario("Carlos")
```

A Importância do return

Explique a diferença entre apenas imprimir (print) dentro da função e devolver um valor para o programa principal com return.

```python
def somar(a, b):
    resultado = a + b
    return resultado

# Guardando o retorno da função em uma variável
total = somar(10, 15)
print(f"O resultado da soma é: {total}")
```

## Exercícios Práticos

Exercício 1: Validador de Maioridade (Fixação de Parâmetros e Retorno Boolean)

    Enunciado: Crie uma função chamada eh_maior_de_idade(idade) que receba uma idade como parâmetro e retorne True se a idade for maior ou igual a 18, ou False caso contrário. No programa principal, peça a idade do usuário e use a função para exibir uma mensagem correspondente.

```python
def eh_maior_de_idade(idade):
    if idade >= 18:
        return True
    else:
        return False

# Programa Principal
idade_usuario = int(input("Digite sua idade: "))

if eh_maior_de_idade(idade_usuario):
    print("Acesso liberado: Você é maior de idade.")
else:
    print("Acesso negado: Você é menor de idade.")
```

Exercício 2: Calculadora de Área de Retângulo (Lógica Matemática Simples)

    Enunciado: Escreva uma função chamada calcular_area(largura, comprimento) que receba as dimensões de um terreno retangular e retorne a sua área total em metros quadrados.

```python
def calcular_area(largura, comprimento):
    area = largura * comprimento
    return area

# Programa Principal
print("=== CÁLCULO DE TERRENO ===")
larg = float(input("Digite a largura do terreno (m): "))
comp = float(input("Digite o comprimento do terreno (m): "))

area_terreno = calcular_area(larg, comp)
print(f"A área total do terreno é de {area_terreno:.2f} m².")
```

Exercício 3: Módulos da Calculadora (Transformando a Aula 3 em Funções)

    Enunciado: Crie quatro funções: somar(a, b), subtrair(a, b), multiplicar(a, b) e dividir(a, b). A função de divisão deve verificar se o divisor é zero antes de calcular para evitar erros. Em seguida, peça dois números ao usuário e execute as quatro operações usando suas funções.

```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return "Erro: Divisão por zero não permitida!"
    return a / b

# Programa Principal
n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))

print("\n=== RESULTADOS ===")
print(f"Soma: {somar(n1, n2)}")
print(f"Subtração: {subtrair(n1, n2)}")
print(f"Multiplicação: {multiplicar(n1, n2)}")
print(f"Divisão: {dividir(n1, n2)}")
```
