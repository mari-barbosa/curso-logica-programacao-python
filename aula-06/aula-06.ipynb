# Aula 06: Automatizando Tarefas com Loops (while)

Por que Repetir Código?

A Abordagem Ineficiente: Se precisarmos mostrar os números de 1 a 5, até poderíamos usar cinco comandos print(). Mas e se fossem 1.000 números?

O Conceito de Loop: Estruturas de repetição executam um mesmo bloco de código enquanto uma condição for verdadeira (True).

A Estrutura do while

Tradução direta: while = "Enquanto".

Três elementos obrigatórios de um laço saudável:

    Inicialização: Criar a variável de controle antes do loop.

    Condição: O teste lógico do while.

    Atualização: Modificar a variável dentro do loop para evitar o loop infinito (quando o programa trava tentando rodar para sempre).

```python
contador = 1  # 1. Inicialização

while contador <= 5:  # 2. Condição
    print(f"Passo número {contador}")
    contador = contador + 1  # 3. Atualização (ou contador += 1)

print("Fim do loop!")
```

Condições de Parada com input()

Mostrar que o while não serve só para contar números, mas também para manter o programa aberto até o usuário decidir sair:

```python
resposta = ""
while resposta.lower() != "sair":
    resposta = input("Digite algo (ou 'sair' para encerrar): ")
    print(f"Você digitou: {resposta}")
```

## Exercícios Práticos

Exercício 1: Contagem Regressiva (Fixação de Contadores Decrementais)

    Enunciado: Escreva um programa que faça uma contagem regressiva para o lançamento de um foguete, de 10 até 0, e exiba "Fogo!" no final.

```python
contador = 10

while contador >= 0:
    print(contador)
    contador -= 1  # Equivalente a: contador = contador - 1

print("🚀 Fogo!")
```

Exercício 2: Somador Infinito (Loop com Condição de Parada)

    Enunciado: Crie um programa que peça para o usuário digitar vários números inteiros. O programa deve somar esses números e só parar quando o usuário digitar o número 0. No final, exiba a soma total.

```python
soma = 0
numero = int(input("Digite um número para somar (ou 0 para sair): "))

while numero != 0:
    soma += numero
    numero = int(input("Digite outro número (ou 0 para sair): "))

print(f"A soma total de todos os números digitados é: {soma}")
```

Exercício 3: O Jogo da Adivinhação (Desafio Prático Inteiro)

    Enunciado: Defina um número secreto fixo no programa (ex: 7). Peça ao usuário para adivinhar esse número. Enquanto ele errar, exiba "Tente novamente!". Quando ele acertar, mostre uma mensagem de parabéns e o total de tentativas que ele precisou.

```python
numero_secreto = 7
palpite = 0
tentativas = 0

print("=== JOGO DA ADIVINHAÇÃO ===")

while palpite != numero_secreto:
    palpite = int(input("Adivinhe o número secreto (1 a 10): "))
    tentativas += 1

    if palpite != numero_secreto:
        print("Número incorreto! Tente novamente.")

print(f"Parabéns! Você acertou o número {numero_secreto} em {tentativas} tentativa(s)!")
```
