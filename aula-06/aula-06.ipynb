## Por que Repetir Código?

```python
from typing import cast

print(1)
print(2)
print(3)
print(4)
print(5)
```

Mas e se precisássemos fazer isso até o número 1000? 

Escreveríamos 1000 linhas de código?

### NÃO! Para isso existem os Loops

Loops são estruturas de repetição que executam um mesmo bloco de código enquanto uma condição for verdadeira (True).

## A Estrutura while (enquanto)

- Três elementos obrigatórios de um laço saudável:

    1. **Inicialização**: Criar a variável de controle antes do loop.

    2. **Condição**: O teste lógico do while.

    3. **Atualização**: Modificar a variável dentro do loop.

- Se você não se atentar a isso, vai criar o temido **LOOP INFINITO**.

```python
contador = 1  # 1. Inicialização

while contador <= 5:  # 2. Condição
    print(f"Passo número {contador}")
    contador = contador + 1  # 3. Atualização (ou contador += 1)

print("Fim do loop!")
```

## Condições de Parada com input()

Todo while pode ser interrompido automaticamente se for combinado com um input( ) específico.

```python
resposta = ""

while resposta.lower() != "sair":
    resposta = input("Digite algo (ou 'sair' para encerrar): ")
    print(f"Você digitou: {resposta}")
```
