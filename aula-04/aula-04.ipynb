## Escrevendo o Primeiro if/else e a Regra da Identação

### Instruções if

- if é a instrução condicional mais simples de todas. Significa, literalmente, **se**.

- Ela permite verificar se algo é verdadeiro ou falso.

```python
x = 1

if x > 0:
    print("x é positivo!")
```

**DUAS COISAS IMPORTANTES:**
- if sempre termina com " : "
- O código que deve ser executado dentro do if precisa ficar **indentado**, com recuo para a direita. Isso indica para o Python que aquela instrução pertence ao bloco do if. Você pode usar a tecla Tab para indentar, e a maioria dos editores faz isso automaticamente.

### Cláusula else

- else em inglês significa, literalmente, "se não".
- Funciona como a segunda parte do if.
- Se a primeira condição for verdadeira, a instrução indentada será executada; **se não** a segunda instrução será executada.

```python
idade = int(input("Digite a sua idade: "))

if idade >= 18:
    print("Você é maior de idade.")
    print("Pode tirar a carteira de motorista!")
else:
    print("Você é menor de idade.")
    print("Ainda não pode dirigir.")

print("Obrigado por usar nosso sistema!")
```
