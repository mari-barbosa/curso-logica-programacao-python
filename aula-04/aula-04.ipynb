## Operadores de Comparação

> : Maior que

< : Menor que

>= : Maior ou igual a

<= : Menor ou igual a

== : Igual a (reforce: dois sinais de igual comparam; um sinal atribui!)

!= : Diferente de

## Escrevendo o Primeiro if/else e a Regra da Identação

A Regra dos Dois Pontos (:): Mostre que o : no final da linha do if avisa ao Python que um bloco de código vai começar.

Identação (4 Espaços/Tab): Destaque que o código dentro do if precisa estar recuado. Se alinhar na margem, o Python dá erro de sintaxe.

```python
idade = int(input("Digite a sua idade: "))

if idade >= 18:
    print("Você é maior de idade.")
    print("Pode tirar a carteira de motorista!")
else:
    print("Você é menor de idade.")
    print("Ainda não pode dirigir.")

print("Obrigado por usar nosso sistema!")  # Linha fora do bloco (executa sempre)
```

## Exercícios Práticos

### Exercício 1: Radar de Velocidade (Fixação do if/else Básico)

    Enunciado: Crie um programa que leia a velocidade de um carro. Se ultrapassar 80 km/h, exiba uma mensagem dizendo que ele foi multado. Caso contrário, exiba uma mensagem dizendo que ele está dentro do limite de velocidade.

```python
velocidade = float(input("Qual é a velocidade atual do carro em km/h? "))

if velocidade > 80:
    print("MULTADO! Você ultrapassou o limite permitido de 80 km/h.")
else:
    print("Tenha um bom dia! Dirija com segurança.")
```

Exercício 2: Par ou Ímpar? (Uso do Operador de Módulo %)

    Enunciado: Desenvolva um programa que receba um número inteiro do usuário e mostre na tela se ele é PAR ou ÍMPAR. (Dica: Todo número par dividido por 2 tem resto igual a 0).

```python
numero = int(input("Digite um número inteiro qualquer: "))

if numero % 2 == 0:
    print(f"O número {numero} é PAR.")
else:
    print(f"O número {numero} é ÍMPAR.")
```

Exercício 3: Validador de Senha (Comparação de Texto)

    Enunciado: Escreva um programa que peça ao usuário para digitar uma senha. Defina uma senha secreta fixa no código (ex: "python123"). Se o usuário acertar, exiba "Acesso Concedido". Se errar, exiba "Acesso Negado".

```python
senha_secreta = "python123"
senha_digitada = input("Digite a senha de acesso: ")

if senha_digitada == senha_secreta:
    print("Acesso Concedido! Bem-vindo ao sistema.")
else:
    print("Acesso Negado! Senha incorreta.")
```
