# Aula 07: Repetições com Limites Definidos (for)

A Diferença Fundamental: while vs for

while (Enquanto): Usado quando não sabemos o momento exato de parar (ex: "Repita enquanto a senha estiver errada").

for (Para cada): Usado quando temos um número fixo de repetições ou uma sequência bem delimitada (ex: "Repita isso para cada número de 1 a 10").

Desmistificando a Função range()

O que é? A função range() gera uma sequência de números inteiros para o for percorrer.

O Limite Superior Exclusivo: Explique com bastante calma que o Python não inclui o último número da contagem.

    range(5) -> Gera: 0, 1, 2, 3, 4 (5 números no total, começando do zero).

    range(1, 6) -> Gera: 1, 2, 3, 4, 5 (do primeiro ao número antes do último).

    range(0, 10, 2) -> Gera: 0, 2, 4, 6, 8 (do 0 ao 10, pulando de 2 em 2).

Escrevendo o Primeiro Loop for

```python
for i in range(1, 4):
    print(f"Rodada {i}")

# Saída:
# Rodada 1
# Rodada 2
# Rodada 3
```

## Exercícios Práticos

Exercício 1: Gerador de Tabuada (Fixação de for e range)

    Enunciado: Escreva um programa que peça um número inteiro ao usuário e mostre a tabuada completa desse número (de 1 a 10) na tela.

```python
numero = int(input("Digite um número para ver sua tabuada: "))

print(f"=== TABUADA DO {numero} ===")
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i:2d} = {resultado}")
```

Exercício 2: Contagem de Números Pares (Uso do Passo em range)

    Enunciado: Crie um programa que mostre na tela todos os números pares situados no intervalo entre 1 e 50.

```python
print("Números pares entre 1 e 50:")
for num in range(2, 51, 2):
    print(num, end=" ")
print("\nFim do programa!")
```

Exercício 3: Acumulador com for (Calculando a Média da Turma)

    Enunciado: Escreva um programa que pergunte quantas notas o professor quer cadastrar. Em seguida, usando o laço for, peça cada uma das notas, calcule a soma delas e exiba a média final da turma.

```python
qtd_notas = int(input("Quantas notas você deseja cadastrar? "))
soma_notas = 0.0

for i in range(1, qtd_notas + 1):
    nota = float(input(f"Digite a nota {i}: "))
    soma_notas += nota

media = soma_notas / qtd_notas
print(f"\nA média final da turma foi: {media:.2f}")
```
